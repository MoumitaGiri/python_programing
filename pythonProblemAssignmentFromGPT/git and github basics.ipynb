#### Git:
##### Git commands:

To config git user:<br>
 **git config --user.name "enter username"** <br>
 **git config --user.email "enter email"**

To check the list of configuration: <br>
**git config --list** <br>

To clone code from master into local: <br>
**git clone "github http project URL"** <br>

To check list of files: <br>
**ls** is the command <br>
To check list of hidden files: <br>
 **ls -a** is the command <br>

To check the current branch: <br>
 **git branch** <br>

To check the status: <br>
 **git status** <br>
Note: ***there are 3 stages of status:*** <br>
1) Change in existing file (**modified**) status <br>
2) New file created not in git (**untracked**) status <br>
3) when both modified and untracked files added then it goes to (**staged**) status means file is ready to be committed <br>
4) after that when we do commit changes then it goes to (**unchanged**) status <br>
 

To add changes: <br>
**git add .** <br>

To commit changes: <br>
**git commit -m "enter some messages"** <br>

To push the changes: <br>
 **git push origin "branch name"** (origin is repo, and master is branch name) <br>
 **git push --set-upstream origin "local branch name"** <br>
 **git push -u origin main** <br>

To work with Init Command: <br>
 **git init** <br>
 **git remote add origin <-link->** <br>
 **git remote -v (to verify remote)** <br>
 **git branch (to check branch)** <br>
 **git branch -M main (to rename branch)** <br>
 **git push origin main** <br>

To work with Branch: <br>
 **git branch** <br>
 **git branch -M "new branch name"** (rename of branch) <br>
 **git checkout -b "new branch name"** (create new branch) <br>
 **git checkout "branch name"** (to nagivate to that brnach) <br>
 **git branch -d "branch name"** (to delete that branch) <br>

To work with merging code: <br>
way1: <br>
 **git diff <"branch name">** (master or main) <br>
 **git merge** (main or master) <br>
way2: <br>
Create a PR (pull request) <br>
 Create PR > review it > merge to master (normal process) <br>

Pull Command: <br>
Used to fetch and download content from a remote repo and immediately upadte the local repo to match that content. <br>
 **git pull origin main** <br>

Resolving Merge conflict: <br>
An event that takes place when Git is unable to automatically resolve differences in code between two commits. <br>
 **git checkout main** <br>
 **git merge "branch name"** <br>

Undoing Changes: <br>
Case1: staged changes (added but no commited) <br>
 **git reset <-file name->** (reset and back to initally state of file) <br>
 **git reset** <br>
Case2: commited changes (for one commit) <br>
 **git rebase Head~1** <br>
Case3: commited changes(for many commits) <br>
 **git log** <br>
 **git reset <-commit hash->** <br>
 **git reset --hard <-commit hash->** <br>

Fork: <br>
Fork is a rough copy <br>
A fork is a new repository that shares code and visibility settings with the original "upstream" <br>
